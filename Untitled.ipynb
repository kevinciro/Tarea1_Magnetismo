```python
import pandas as pd
from MonteCarlo import graficar_historial, graficar_historial_normalizado, calcular_metricas_desde_csv, graficar_observable_vs_T
```

```python
# Cargar un CSV especifico
df = pd.read_csv("simulacion_L_24_T_1.0_periodica.csv")

# Graficar reutilizando las funciones que ya tienes
graficar_historial(
    df["mcs"],
    df["E_prom"], df["E2_prom"],
    df["M_prom"], df["M2_prom"],
    df["Ms_prom"], df["Ms2_prom"],
)

limite = 5_000

graficar_historial_normalizado(
    df["mcs"][:limite],
    df["E_prom"][:limite], df["E2_prom"][:limite],
    df["M_prom"][:limite], df["M2_prom"][:limite],
    df["Ms_prom"][:limite], df["Ms2_prom"][:limite],
)
```

```python
T_con_tao = [0.1, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0]
tao_libres = [1000, 2000, 2000, 2000, 2000, 2000, 2000, 3000, 2000, 1000, 2000]
tao_periodicas = [1000, 2000, 2000, 3000, 2000, 2000, 1000, 3000, 2000, 3000, 2000]
```

```python
import glob
import re

tao_por_frontera = {
    "libre": dict(zip(T_con_tao, tao_libres)),
    "periodica": dict(zip(T_con_tao, tao_periodicas)),
}
tao_por_defecto = 2000  # para T que no esten en las listas de arriba (ej. T=0.001)

patron = re.compile(r"simulacion_L_(\d+)_T_([\d.]+)_(libre|periodica)\.csv")

resultados_libre = []
resultados_periodica = []

for archivo in sorted(glob.glob("simulacion_L_*_T_*.csv")):
    m = patron.match(archivo)
    if not m:
        continue
    L_archivo = int(m.group(1))
    T_archivo = float(m.group(2))
    frontera = m.group(3)

    tao = tao_por_frontera[frontera].get(T_archivo, tao_por_defecto)
    if T_archivo not in tao_por_frontera[frontera]:
        print(f"Aviso: no hay tao definido para T={T_archivo} ({frontera}), usando el valor por defecto ({tao_por_defecto})")

    df = pd.read_csv(archivo)
    r = calcular_metricas_desde_csv(tao, df["E"].to_numpy(), df["Ms"].to_numpy(), L_archivo, T_archivo)
    r["tao"] = tao

    if frontera == "libre":
        resultados_libre.append(r)
    else:
        resultados_periodica.append(r)

# Se ordenan por L y T para que las curvas queden en orden al graficar despues
resultados_libre = sorted(resultados_libre, key=lambda r: (r["L"], r["T"]))
resultados_periodica = sorted(resultados_periodica, key=lambda r: (r["L"], r["T"]))

print(f"{len(resultados_libre)} resultados libres, {len(resultados_periodica)} resultados periodicos")
```

```python
df_libre = pd.DataFrame(resultados_libre)
df_periodica = pd.DataFrame(resultados_periodica)

# Seccion 3.2.1, punto 1: Ms(T) y chi_s(T) para ambos L y ambas condiciones de frontera
graficar_observable_vs_T(df_libre, df_periodica, "Ms", r"$M_s$", titulo=r"Magnetización de subred $M_s(T)$")
graficar_observable_vs_T(df_libre, df_periodica, "chi_s", r"$\chi_s$", titulo=r"Susceptibilidad de subred $\chi_s(T)$")
```
